```python
import pandas as pd
import numpy as np
import zipfile
from pathlib import Path

print("Libraries imported successfully!")
```

```python
# Find the project root
project_root = Path.cwd()

# VS Code may run notebooks from the notebooks folder
if not (project_root / "data").exists():
    project_root = project_root.parent

dataset_path = project_root / "data" / "raw" / "CMAPSSData.zip"

print("Dataset path:", dataset_path)
print("Dataset exists:", dataset_path.exists())
```

```python
with zipfile.ZipFile(dataset_path, "r") as zip_file:
    filenames = zip_file.namelist()

for filename in filenames:
    print(filename)
```

```python
columns = (
    ["engine_id", "cycle"]
    + [f"setting_{i}" for i in range(1, 4)]
    + [f"sensor_{i}" for i in range(1, 22)]
)

print(columns)
print("Total columns:", len(columns))
```

```python
with zipfile.ZipFile(dataset_path, "r") as zip_file:
    with zip_file.open("train_FD001.txt") as file:
        train_df = pd.read_csv(
            file,
            sep=r"\s+",
            header=None,
            names=columns
        )

print("Training data loaded successfully!")
print(train_df.shape)

display(train_df.head())
```

```python
display(train_df.tail())
```

```python
train_df.info()
```

```python
engine_1 = train_df[
    train_df["engine_id"] == 1
]

print("Engine 1 records:", len(engine_1))
print("First cycle:", engine_1["cycle"].min())
print("Final cycle:", engine_1["cycle"].max())
```

```python
engine_2 = train_df[
    train_df["engine_id"] == 2
]

print("Engine 2 records:", len(engine_2))
print("Final cycle:", engine_2["cycle"].max())
```

```python
# Engine 2 operates longer than Engine 1. This is why we cannot simply assume every engine will fail at the same number of cycles.
# Our future model must analyse sensor measurements to learn the differences between engine histories.
```

```python
with zipfile.ZipFile(dataset_path, "r") as zip_file:

    with zip_file.open("test_FD001.txt") as file:
        test_df = pd.read_csv(
            file,
            sep=r"\s+",
            header=None,
            names=columns
        )

    with zip_file.open("RUL_FD001.txt") as file:
        rul_df = pd.read_csv(
            file,
            sep=r"\s+",
            header=None,
            names=["RUL"]
        )

print("Training:", train_df.shape)
print("Testing:", test_df.shape)
print("Actual RUL:", rul_df.shape)
```

```python
print(
    "Test Engine 1 last cycle:",
    test_df[test_df["engine_id"] == 1]["cycle"].max()
)

print(
    "Test Engine 1 actual RUL:",
    rul_df.iloc[0]["RUL"]
)
```

```python
print("Training engines:", train_df["engine_id"].nunique())
print("Testing engines:", test_df["engine_id"].nunique())

print("Missing training values:", train_df.isnull().sum().sum())
print("Missing testing values:", test_df.isnull().sum().sum())

print("Duplicate training rows:", train_df.duplicated().sum())
```
